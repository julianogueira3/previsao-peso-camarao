# 07 — Conclusões

## O que eu fiz

Parti de um formulário de campo real de uma fazenda de camarão e construí um modelo de
aprendizado de máquina pra prever o **peso** do camarão ao longo do cultivo. Minha ideia
de modelagem, inspirada no estudo de Mujahid et al. (2025) na Indonésia, foi tratar cada
biometria como um exemplo e usar o **dia de cultivo** e a **ração acumulada** como
preditores principais.

## O que eu encontrei

- O modelo Random Forest prevê o peso com erro médio de cerca de **2,4 gramas** e R² em
  torno de **0,68** ou seja, explica cerca de dois terços da variação do peso.
- As variáveis mais importantes são o **dia de cultivo** e a **ração acumulada por
  animal**, exatamente como na literatura que estudei.
- Comparei oito algoritmos e os modelos de árvore (Random Forest, Extra Trees) se saíram
  melhor; modelos mais complexos como XGBoost não trouxeram ganho, o que é esperado pro
  tamanho do meu conjunto de dados.

## Aplicação prática

Na prática, esse modelo pode ajudar o produtor a **estimar quando o camarão vai atingir o
peso de despesca**, planejando melhor a colheita e o manejo alimentar. Sabendo os dias de
cultivo e a ração fornecida, ele consegue uma previsão do peso atual sem precisar pesar.

## Limitações e trabalhos futuros

- Os dados vêm de uma única fazenda, então o modelo pode não generalizar direto pra
  outras.
- Fatores como qualidade da pós-larva e sanidade não estão nos dados e poderiam refinar as
  previsões.
- Um passo futuro seria integrar dados de sensores contínuos pra prever o crescimento em
  tempo real.
