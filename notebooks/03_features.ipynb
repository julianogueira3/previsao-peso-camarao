## A ração acumulada: minha principal escolha de feature

No formulário eu tenho a ração de cada dia (`TDIA`). Mas o que faz o camarão crescer não
é a ração de um dia só é **tudo o que ele comeu até aquele momento**. Então eu criei a
`RACAO_ACUM`: a soma de toda a ração desde o início do cultivo até cada pesagem. E dividi
pela quantidade de animais (`RACAO_ACUM_POR_PL`) pra poder comparar cultivos de tamanhos
diferentes. Essa foi a feature que mais fez diferença depois do tempo e é a mesma lógica
de "cumulative feed" que o artigo da Indonésia usou.
